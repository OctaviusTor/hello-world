# Learning Python Concepts and VS Code

This notebook will guide you through fundamental Python concepts and how to use Visual Studio Code (VS Code) for Python development.

## 1. Setting Up VS Code for Python Development

To get started with Python in VS Code:

1. **Install the Python Extension**: Go to the Extensions view (Ctrl+Shift+X), search for "Python", and install the official Microsoft Python extension.

2. **Configure Python Interpreter**: Open the Command Palette (Ctrl+Shift+P), type "Python: Select Interpreter", and choose your Python environment.

3. **Create a Python File**: Create a new file with `.py` extension and start coding. VS Code will provide IntelliSense, syntax highlighting, and debugging support.

## 2. Basic Python Syntax and Variables

Python uses indentation for code blocks. Variables are assigned with `=`. Let's see some basics:

```python
# Variable assignment
name = "Charles"
age = 52
height = 6.2

# Print statements
print("Hello, " + name + "!")
print(f"You are {age} years old and {height} feet tall.")

# Basic arithmetic
x = 10
y = 3
print(f"x + y = {x + y}")
print(f"x - y = {x - y}")
print(f"x * y = {x * y}")
print(f"x / y = {x / y}")
```

## 3. Data Types and Operations

Python has several built-in data types: integers, floats, strings, booleans, lists, tuples, dictionaries, and sets.

```python
# Integers and floats
num_int = 42
num_float = 3.14
print(f"Integer: {num_int}, type: {type(num_int)}")
print(f"Float: {num_float}, type: {type(num_float)}")

# Strings
text = "Hello, World!"
print(f"String: {text}, length: {len(text)}")

# Booleans
is_true = True
is_false = False
print(f"Boolean: {is_true}, type: {type(is_true)}")

# Lists (mutable sequences)
fruits = ["apple", "banana", "cherry"]
fruits.append("date")
print(f"List: {fruits}")

# Tuples (immutable sequences)
coordinates = (10, 20)
print(f"Tuple: {coordinates}")

# Dictionaries (key-value pairs)
person = {"name": "Bob", "age": 30, "city": "New York"}
print(f"Dictionary: {person}")
print(f"Bob's age: {person['age']}")

# Sets (unordered unique elements)
unique_numbers = {1, 2, 3, 3, 2}
print(f"Set: {unique_numbers}")
```

## 4. Control Structures: Loops and Conditionals

Control structures allow you to make decisions and repeat actions in your code.

```python
# If-else statements
age = 18
if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")

# For loops
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"I like {fruit}")

# While loops
count = 0
while count < 5:
    print(f"Count: {count}")
    count += 1

# List comprehension (compact way to create lists)
squares = [x**2 for x in range(1, 6)]
print(f"Squares: {squares}")

# Conditional in list comprehension
even_squares = [x**2 for x in range(1, 11) if x % 2 == 0]
print(f"Even squares: {even_squares}")
```

## 5. Defining and Using Functions

Functions are reusable blocks of code that perform specific tasks. They help organize and modularize your code.

```python
# Simple function
def greet(name):
    return f"Hello, {name}!"

print(greet("Alice"))

# Function with multiple parameters
def add_numbers(a, b):
    return a + b

result = add_numbers(5, 3)
print(f"5 + 3 = {result}")

# Function with default parameters
def power(base, exponent=2):
    return base ** exponent

print(f"2^3 = {power(2, 3)}")
print(f"5^2 = {power(5)}")  # Uses default exponent=2

# Function with variable arguments
def sum_all(*args):
    return sum(args)

print(f"Sum of 1, 2, 3, 4, 5: {sum_all(1, 2, 3, 4, 5)}")

# Lambda functions (anonymous functions)
square = lambda x: x ** 2
print(f"Square of 7: {square(7)}")

# Using functions in higher-order functions
numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(lambda x: x**2, numbers))
print(f"Squared numbers: {squared_numbers}")
```

## 7. Using VS Code's Integrated Terminal

VS Code has a built-in terminal that allows you to run commands without leaving the editor. You can run Python scripts, install packages, and more.

To use the integrated terminal:

1. Open the terminal: View > Terminal (or Ctrl+`)
2. Run Python scripts: `python script.py`
3. Install packages: `pip install package_name`
4. Check Python version: `python --version`
5. Run interactive Python: `python` (then type code and press Enter)

You can also run commands directly from the editor by right-clicking and selecting "Run Python File in Terminal".

## 8. Debugging Python Code in VS Code

VS Code's debugger allows you to step through your code, inspect variables, and find bugs.

```python
# Example code to debug
def calculate_sum(numbers):
    total = 0
    for num in numbers:
        total += num
        print(f"Adding {num}, total now: {total}")  # Add breakpoint here
    return total

numbers = [1, 2, 3, 4, 5]
result = calculate_sum(numbers)
print(f"Final result: {result}")

# To debug:
# 1. Click to the left of the line number to add a breakpoint (red dot)
# 2. Press F5 or go to Run > Start Debugging
# 3. Use F10 to step over, F11 to step into
# 4. Watch variables in the Variables panel
```
